```python
import pandas as pd

# Load the CSV file with encoding detection
file_path = './data/demo_data/football/Forwards.csv'

try:
    df = pd.read_csv(file_path, encoding='utf-8')
except UnicodeDecodeError:
    try:
        df = pd.read_csv(file_path, encoding='latin-1')
    except UnicodeDecodeError:
        df = pd.read_csv(file_path, encoding='iso-8859-1')

# Check available players
print("Available players in dataset:")
print(df.iloc[:, 0].unique())  # Assuming first column is player name
print("\n" + "="*60 + "\n")

# Use flexible player matching pattern
player_name = "Pedro"
parts = player_name.split()
pattern = ".*".join(parts)

pedro_filtered = df[df.iloc[:, 0].str.contains(pattern, case=False, na=False, regex=True)]

if pedro_filtered.empty:
    print(f"ERROR: '{player_name}' not found in dataset. Check the player name above.")
else:
    pedro = pedro_filtered.iloc[0]
    
    # Display the player's stats
    print(f"{player_name} Statistics:")
    print(pedro)
    print("\n" + "="*60 + "\n")
    
    # EDA - Describe strengths and weaknesses
    print(f"EXPLORATORY DATA ANALYSIS - {player_name}\n")
    
    # Get numeric columns for analysis (exclude player name column)
    numeric_cols = df.select_dtypes(include=['number']).columns
    
    # Find above-average stats (strengths) and below-average stats (weaknesses)
    above_avg = []
    below_avg = []
    
    for col in numeric_cols:
        player_value = pedro[col]
        dataset_mean = df[col].mean()
        
        if player_value > dataset_mean * 1.2:  # 20% above average
            above_avg.append((col, player_value))
        elif player_value < dataset_mean * 0.8:  # 20% below average
            below_avg.append((col, player_value))
    
    # First sentence - Strengths
    if above_avg:
        strengths_text = f"{player_name} excels in " + ", ".join([col for col, _ in above_avg]) + "."
    else:
        strengths_text = f"{player_name} demonstrates balanced performance across metrics."
    print(f"{strengths_text}\n")
    
    # Second sentence - Average and weak areas
    if below_avg:
        weak_text = f"This player shows average to below-average performance in " + ", ".join([col for col, _ in below_avg]) + "."
    else:
        weak_text = "This player maintains consistent above-average stats throughout."
    print(f"{weak_text}\n")
    
    # Concluding statement
    print(f"Conclusion: {player_name} is a forward with distinctive strengths in key attacking metrics, offering valuable contributions to team performance in specific tactical areas.")
```
