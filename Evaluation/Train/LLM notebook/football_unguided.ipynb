```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Forwards.csv')

# Define the metrics to be analyzed
metrics = [
    'npxG_adjusted_per90', 'final_third_passes_adjusted_per90',
    'final_third_receptions_adjusted_per90', 'ground_duels_won_adjusted_per90',
    'air_duels_won_adjusted_per90', 'smart_passes_adjusted_per90',
    'goals_adjusted_per90', 'assists_adjusted_per90', 'key_passes_adjusted_per90'
]

# Extract Pedro's specific data
pedro_data = df[df['shortName'] == 'Pedro'].iloc[0]

# Calculate the percentile rank for each metric to understand his relative standing
percentiles = {}
for metric in metrics:
    # Percentage of players that Pedro outperforms in this metric
    percentiles[metric] = (df[metric] < pedro_data[metric]).mean() * 100

# Display the results for analysis
print(f"Analysis for {pedro_data['shortName']}:")
for metric in metrics:
    print(f"{metric:40} | Value: {pedro_data[metric]:.3f} | Percentile: {percentiles[metric]:.1f}%")

# Calculate dataset averages for additional context
print("\nDataset Averages:")
print(df[metrics].mean())
```
