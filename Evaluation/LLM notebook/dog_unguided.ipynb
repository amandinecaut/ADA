```python
import pandas as pd

# Load the dataset
df = pd.read_csv('breed_traits.csv')

# Clean breed names to handle non-breaking spaces and whitespace
df['Breed_Clean'] = df['Breed'].apply(lambda x: x.replace('\xa0', ' ').strip())

# Filter for Great Pyrenees
gp_stats = df[df['Breed_Clean'] == 'Great Pyrenees'].iloc[0]

# Calculate dataset averages for comparison
numeric_cols = df.select_dtypes(include=['number']).columns
dataset_means = df[numeric_cols].mean()

# Identify Strengths (Score of 5)
strengths = gp_stats[numeric_cols][gp_stats[numeric_cols] == 5].index.tolist()

# Identify Average/Weaknesses (Scores below or equal to 3)
# Note: For some traits like Shedding/Drooling, a high score is a maintenance "weakness"
average_or_below = gp_stats[numeric_cols][gp_stats[numeric_cols] <= 3].index.tolist()

print("--- Great Pyrenees EDA ---")
print(f"Top Strengths: {', '.join(strengths)}")
print("\nSpecific Scores:")
print(gp_stats[numeric_cols])

print("\nComparison to Dataset Averages:")
comparison = pd.DataFrame({
    'Great Pyrenees': gp_stats[numeric_cols],
    'Dataset Average': dataset_means.round(2)
})
print(comparison)
```
