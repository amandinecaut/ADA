```python
import pandas as pd

# Load the CSV file
csv_path = 'C:/Users/amaca253/Documents/datomatisation/datomatisation-dev/data/demo_data/dogs/breed_traits.csv'
df = pd.read_csv(csv_path)

# Check available breeds
print("Available breeds in dataset:")
print(df['Breed'].unique())
print("\n" + "="*60 + "\n")

# Use flexible breed matching pattern
breed_name = "Great Pyrenees"
parts = breed_name.split()
pattern = ".*".join(parts)

great_pyrenees_filtered = df[df['Breed'].str.contains(pattern, case=False, na=False, regex=True)]

if great_pyrenees_filtered.empty:
    print(f"ERROR: '{breed_name}' not found in dataset. Check the breed name above.")
else:
    great_pyrenees = great_pyrenees_filtered.iloc[0]
    
    # Display the breed's traits
    print(f"The {breed_name} Traits:")
    print(great_pyrenees)
    print("\n" + "="*60 + "\n")
    
    # EDA - Describe strengths and weaknesses
    print(f"EXPLORATORY DATA ANALYSIS - {breed_name}\n")
    
    # Get numeric columns for analysis
    numeric_cols = df.select_dtypes(include=['number']).columns
    
    # Find above-average traits (strengths)
    above_avg = []
    below_avg = []
    
    for col in numeric_cols:
        breed_value = great_pyrenees[col]
        dataset_mean = df[col].mean()
        
        if breed_value > dataset_mean * 1.2:  # 20% above average
            above_avg.append((col, breed_value))
        elif breed_value < dataset_mean * 0.8:  # 20% below average
            below_avg.append((col, breed_value))
    
    # First sentence - Strengths
    if above_avg:
        strengths_text = f"The {breed_name} excels in " + ", ".join([col for col, _ in above_avg]) + "."
    else:
        strengths_text = f"The {breed_name} demonstrates balanced performance across traits."
    print(f"{strengths_text}\n")
    
    # Second sentence - Average and weak areas
    if below_avg:
        weak_text = f"This breed shows average to below-average performance in " + ", ".join([col for col, _ in below_avg]) + "."
    else:
        weak_text = "This breed maintains consistent above-average traits throughout."
    print(f"{weak_text}\n")
    
    # Concluding statement
    print(f"Conclusion: The {breed_name} is a strong guardian breed with notable protective and physical capabilities, though it requires specific training approaches for optimal performance in certain behavioral areas.")
```
