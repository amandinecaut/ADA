```python
import numpy as np
import pandas as pd
from factor_analyzer import FactorAnalyzer
from factor_analyzer.utils import polychoric_correlation
```

```python
class PolychoricFAStrategy:
    name = "Polychoric FA"

    def fit(self, df, n_factors):
        # Polychoric correlation expects ordinal integers
        corr, _ = polychoric_correlation(df.values)
        model = FactorAnalyzer(
            n_factors=n_factors,
            rotation="varimax",
            is_corr_matrix=True
        )
        model.fit(corr)
        return model
        
from abc import ABC, abstractmethod

class FactorStrategy(ABC):
    name: str

    @abstractmethod
    def fit(self, df: pd.DataFrame, n_factors: int):
        pass

import prince

class MCAStrategy(FactorStrategy):
    name = "MCA"

    def fit(self, df, n_factors):
        model = prince.MCA(n_components=n_factors, random_state=42)
        model.fit(df)
        return model

from sklearn.decomposition import FactorAnalysis

class ContinuousFAStrategy(FactorStrategy):
    name = "FA"

    def fit(self, df, n_factors):
        model = FactorAnalysis(n_components=n_factors)
        model.fit(df)
        return model
        
class PolychoricFAStrategy(FactorStrategy):
    name = "Polychoric FA"

    def fit(self, df, n_factors):
        corr, _ = polychoric_correlation(df.values)
        model = FactorAnalyzer(
            n_factors=n_factors,
            rotation="varimax",
            is_corr_matrix=True
        )
        model.fit(corr)
        return model

class FAMDStrategy(FactorStrategy):
    name = "FAMD"

    def fit(self, df, n_factors):
        df = df.copy()
        for col in df.select_dtypes(include=[np.number]).columns:
            df[col] = df[col].astype(float)

        model = prince.FAMD(n_components=n_factors, random_state=42)
        model.fit(df)
        return model
```

```python
def select_strategy(df: pd.DataFrame) -> FactorStrategy:
    num_cols = df.select_dtypes(include=[np.number]).columns
    cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns

    if len(num_cols) == 0:
        return MCAStrategy()

    ordinal_cols = [c for c in num_cols if df[c].nunique() < 10]

    if len(cat_cols) == 0 and len(ordinal_cols) == len(num_cols):
        return PolychoricFAStrategy()

    if len(cat_cols) == 0:
        return ContinuousFAStrategy()

    return FAMDStrategy()

def perform_best_factor_analysis(df, n_factors):
    strategy = select_strategy(df)
    model = strategy.fit(df, n_factors)
    return model, strategy.name
```

```python
from factor_analyzer.factor_analyzer import (
    calculate_kmo,
    calculate_bartlett_sphericity
)

def check_factor_adequacy(df: pd.DataFrame, alpha=0.05):
    """
    Returns a dict with KMO and Bartlett test results.
    """
    chi2, p_value = calculate_bartlett_sphericity(df)
    kmo_all, kmo_model = calculate_kmo(df)

    return {
        "kmo": kmo_model,
        "bartlett_chi2": chi2,
        "bartlett_p_value": p_value,
        "is_factorable": (kmo_model >= 0.6) and (p_value < alpha)
    }
    
def select_rotation(corr_matrix, threshold=0.3):
    """
    Decide between varimax (orthogonal) and oblimin (oblique).
    """
    off_diag = corr_matrix - np.eye(corr_matrix.shape[0])
    max_corr = np.abs(off_diag).max()

    return "oblimin" if max_corr > threshold else "varimax"
```

```python
from factor_analyzer import FactorAnalyzer
import pingouin as pg

class PolychoricFAStrategy(FactorStrategy):
    name = "Polychoric FA"

    def fit(self, df, n_factors):
        # Adequacy checks
        adequacy = check_factor_adequacy(df)
        if not adequacy["is_factorable"]:
            raise ValueError(f"Data not factorable: {adequacy}")

        # Polychoric correlation
        corr, _ = polychoric_correlation(df.values)

        # Initial FA to estimate factor correlations
        temp_fa = FactorAnalyzer(
            n_factors=n_factors,
            rotation=None,
            is_corr_matrix=True
        )
        temp_fa.fit(corr)

        # Rotation decision
        rotation = select_rotation(
            temp_fa.get_factor_correlation_matrix()
            if temp_fa.get_factor_correlation_matrix() is not None
            else np.eye(n_factors)
        )

        # Final FA
        model = FactorAnalyzer(
            n_factors=n_factors,
            rotation=rotation,
            is_corr_matrix=True
        )
        model.fit(corr)

        # Attach diagnostics
        model.adequacy_ = adequacy
        model.rotation_ = rotation

        return model
```

```python
def export_loadings(model, feature_names, factor_prefix="Factor"):
    """
    Returns a tidy DataFrame of factor loadings.
    """
    loadings = model.loadings_

    factors = [
        f"{factor_prefix}{i+1}"
        for i in range(loadings.shape[1])
    ]

    df_loadings = (
        pd.DataFrame(loadings, index=feature_names, columns=factors)
        .reset_index()
        .rename(columns={"index": "variable"})
        .melt(id_vars="variable", var_name="factor", value_name="loading")
        .sort_values(by=["factor", "loading"], ascending=[True, False])
    )

    return df_loadings
```
